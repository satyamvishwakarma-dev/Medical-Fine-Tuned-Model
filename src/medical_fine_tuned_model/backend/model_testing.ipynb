```
import os
from transformers import AutoTokenizer, AutoModelForCausalLM

# 1. Configuration
# Replace with your target model (e.g., "meta-llama/Llama-3-8B", "mistralai/Mistral-7B-v0.1", "gpt2")
MODEL_ID = "google/gemma-3-1b-it"
SAVE_DIRECTORY = "./model_for_finetuning"

print(f"🔄 Initialising download for {MODEL_ID}...")

# 2. Download Tokenizer and Model
# This automatically handles the tqdm progress bar, showing size and speed.
try:
    print("Downloading tokenizer configurations...")
    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_ID,
        cache_dir=SAVE_DIRECTORY,
        use_fast=True
    )

    print("Downloading model weights and config files...")
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        cache_dir=SAVE_DIRECTORY,
        device_map="cpu" # Kept on CPU during download to save GPU VRAM
    )

    print("\nDownload completed successfully!")

    # 3. Verify files on disk
    print("\nVerified local assets for training:")
    for root, dirs, files in os.walk(SAVE_DIRECTORY):
        for file in files:
            file_path = os.path.join(root, file)
            size_mb = os.path.getsize(file_path) / (1024 * 1024)
            print(f" -> {file} ({size_mb:.2f} MB)")

except Exception as e:
    print(f"\n❌ Download failed or interrupted: {e}")
    print("💡 Tip: Re-running this script will resume the download from where it stopped.")
```
